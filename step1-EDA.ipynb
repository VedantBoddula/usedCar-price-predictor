#  step 2-Split into training and test sets


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    dataFeatures,
    dataLabels,
    test_size=0.2,
    random_state=42
)

```

```python
# X_train → 80% of the FEATURES
# y_train → 80% of the LABELS

# X_test  → 20% of the FEATURES
# y_test  → 20% of the LABELS

#During Training model.fit(X_train, y_train)

#During testing predictions = model.predict(X_test)
```

```python
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
num_attributes = X_train.select_dtypes(include=["int64","float64"]).columns.tolist()
```

```python
cat_attributes = X_train.select_dtypes(include=["object"]).columns.tolist()
```

```python
num_attributes
```

```python
cat_attributes
```

# step 4-Now create the numerical & Categorical Pipeline:

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("standard_scaler", StandardScaler())
])

print(num_pipeline)
```

```python
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("one_hot_encoder", OneHotEncoder(handle_unknown="ignore"))
])

print(cat_pipeline)
```

# Step 6: combine the numerical and categorical pipelines.

```python
full_pipeline = ColumnTransformer([
    ("num",num_pipeline, num_attributes),
    ("cat",cat_pipeline, cat_attributes)
])
print(full_pipeline)
```

# Step 7 — Transform the training data

```python
data_prepared = full_pipeline.fit_transform(X_train)

print(data_prepared.shape)
print(data_prepared)
```

# step 9: evaluate  all the models using cross-validation by checking their rmse value (lower rmse = Better performance)

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score

lin_reg = LinearRegression()

lin_rmse = -cross_val_score(
    lin_reg,
    data_prepared,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=10
)

print("Linear Regression RMSE:")
print(pd.Series(lin_rmse).describe())
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_reg = DecisionTreeRegressor(random_state=42)
tree_rmse = -cross_val_score(
    tree_reg,
    data_prepared,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=10
)

print("Decision Tree RMSE:")
print(pd.Series(tree_rmse).describe())
```

```python
from sklearn.ensemble import RandomForestRegressor

forest_reg = RandomForestRegressor(random_state=42)
forest_rmse = -cross_val_score(
    forest_reg,
    data_prepared,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=10
)

print("Random Forest RMSE:")
print(pd.Series(forest_rmse).describe())
```

#  step -10 select the model and train the model as linReg rmse value is less so the model performs well

```python
lin_reg.fit(data_prepared, y_train)

print("Linear Regression trained on the full training set.")
```

```python
X_test_prepared = full_pipeline.transform(X_test)

print(X_test_prepared.shape)
```

```python
predictions = lin_reg.predict(X_test_prepared)

print(predictions[:10])
```

```python
from sklearn.metrics import root_mean_squared_error

test_rmse = root_mean_squared_error(y_test, predictions)

print("Final Test RMSE:", test_rmse)
```
